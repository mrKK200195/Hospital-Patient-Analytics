# Data Loading & Initial Inspection

Set up the notebook and locate the raw hospital datasets.

```python
# DATA LOADING & INITIAL INSPECTION

import pandas as pd
import numpy as np

from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

print("=" * 70)
print("HOSPITAL PATIENT ANALYTICS - DATA LOADING")
print("=" * 70)


# Define project and data paths
PROJECT_ROOT = Path.cwd().parent
RAW_PATH = PROJECT_ROOT / "data" / "raw"

print("\nProject root:")
print(PROJECT_ROOT)

print("\nRaw data folder:")
print(RAW_PATH)
```

## 2. Check Required Files

Confirm that all expected raw CSV files are present.

```python
files = {
    "Patients": "patients.csv",
    "Services Weekly": "services_weekly.csv",
    "Staff": "staff.csv",
    "Staff Schedule": "staff_schedule.csv"
}

print("\n" + "=" * 70)
print("CHECKING FILES")
print("=" * 70)

for name, filename in files.items():
    file_path = RAW_PATH / filename

    if file_path.exists():
        print(f"✓ {name}: {filename} found")
    else:
        print(f"✗ {name}: {filename} NOT FOUND")
```

## 3. Load Datasets

Read the four source CSV files into pandas DataFrames.

```python
# Load datasets

patients = pd.read_csv(RAW_PATH / "patients.csv")
services = pd.read_csv(RAW_PATH / "services_weekly.csv")
staff = pd.read_csv(RAW_PATH / "staff.csv")
staff_schedule = pd.read_csv(RAW_PATH / "staff_schedule.csv")

print("\n✓ All datasets loaded successfully!")
```

## 4. Organize Datasets

Group the loaded DataFrames in a dictionary for consistent inspection.

```python
# Store datasets in dictionary

datasets = {
    "Patients": patients,
    "Services Weekly": services,
    "Staff": staff,
    "Staff Schedule": staff_schedule
}
```

## 5. Dataset Size

Review the number of rows and columns in each dataset.

```python
# Dataset size

print("\n" + "=" * 70)
print("DATASET SIZE")
print("=" * 70)

for name, df in datasets.items():
    print(
        f"{name:<20} : "
        f"{df.shape[0]:,} rows × {df.shape[1]} columns"
    )
```

## 6. Column Names

List the fields available in each dataset.

```python
# Column names

print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for name, df in datasets.items():
    print(f"\n{name}:")
    print(df.columns.tolist())
```

## 7. First Five Records

Preview the first five rows from each dataset.

```python
# First 5 records

print("\n" + "=" * 70)
print("FIRST 5 RECORDS")
print("=" * 70)

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    display(df.head())
```

## 8. Data Types

Check the inferred data type of every column.

```python
# Data types

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print(df.dtypes)
```

## 9. Missing Values

Identify columns with missing data in each dataset.

```python
# Missing values

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    missing = df.isnull().sum()

    if missing.sum() == 0:
        print("No missing values.")
    else:
        print(missing[missing > 0])
```

## 10. Duplicate Records

Count exact duplicate rows in each dataset.

```python
# Duplicate records

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name:<20} : {duplicates} duplicate rows")
```

## 11. Numerical Statistics

Summarize numeric columns with descriptive statistics.

```python
# Numerical statistics

print("\n" + "=" * 70)
print("NUMERICAL STATISTICS")
print("=" * 70)

for name, df in datasets.items():

    numeric_columns = df.select_dtypes(
        include=np.number
    ).columns

    print(f"\n--- {name} ---")

    if len(numeric_columns) > 0:
        display(df[numeric_columns].describe().T)
    else:
        print("No numerical columns.")
```

## 12. Data Information

Review dataset dimensions and memory usage.

```python
# Data information

print("\n" + "=" * 70)
print("DATA INFORMATION")
print("=" * 70)

for name, df in datasets.items():

    print(f"\n--- {name} ---")

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Memory usage:",
          round(df.memory_usage(deep=True).sum() / 1024, 2),
          "KB")
```

Mark the end of the initial loading and inspection workflow.
